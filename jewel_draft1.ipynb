```python
import requests
import zipfile
import io
import os
import pandas as pd

#Only if the Blue bike file doesn't already exist
if os.path.exists("bluebike_trips.csv"):
    print("bluebike_trips.csv already exists, skipping download.")
else:
    # Download the zip file from Bluebikes S3\n",
    url = "https://s3.amazonaws.com/hubway-data/202411-bluebikes-tripdata.zip",
    print("Downloading data...")
    response = requests.get(url)
    response.raise_for_status()

    # Extract the CSV from the zip in memory\n",
    with zipfile.ZipFile(io.BytesIO(response.content)) as z:
        # List files inside the zip to find the CSV\n",
        csv_files = [f for f in z.namelist() if f.endswith('.csv')]
        #print(f"Found files in zip: {z.namelist()}")
    
    # Read the first (and typically only) CSV
    with z.open(csv_files[0]) as f:
        df = pd.read_csv(f)

#print(f"Loaded {len(df):,} rows and {len(df.columns)} columns")
#print(df.head())

# Save to bluebike_trips.csv
df.to_csv("bluebike_trips.csv", index=False)
#print("Saved to bluebike_trips.csv")
```

```python
#City of Boston 311 Service Requests

#Only if the 311_service_requests file doesn't already exist
if os.path.exists("311_service_requests.csv"):
    print("311_service_requests.csv already exists, skipping download.")
else:
    # Download the zip file from 311 Service Requests
    url = "https://data.boston.gov/dataset/8048697b-ad64-4bfc-b090-ee00169f2323/resource/254adca6-64ab-4c5c-9fc0-a6da622be185/download/tmp7o3lwr4m.csv",
    print("Downloading data...")
    response = requests.get(url)
    response.raise_for_status()
    # Extract the CSV from the zip in memory
    with zipfile.ZipFile(io.BytesIO(response.content)) as z:
        # List files inside the zip to find the CSV
        csv_files = [f for f in z.namelist() if f.endswith('.csv')]
        print(f"Found files in zip: {z.namelist()}")
    
    # Read the first (and typically only) CSV
    with z.open(csv_files[0]) as f:
        df = pd.read_csv(f)

#print(f"Loaded {len(df):,} rows and {len(df.columns)} columns")
#print(df.head())

# Save to 311_service_requests.csv
df.to_csv("311_service_requests.csv", index=False)
#print("Saved to 311_service_requests.csv")
```
